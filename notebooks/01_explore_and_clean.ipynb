# PaperTrail: Data Cleaning and Exploration

**Reading the raw paper leaks CSV to inspect its structure before cleaning.**

```python
import pandas as pd
df = pd.read_csv('../data/raw/paper_leaks.csv')
df.head()
```

**Checking data quality and missing values**

```python
df.info()
```

```python
df.isnull().sum()
```

**Converting date column to proper datetime type**

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

**Normalizing the area/state column: stripping city-level detail for consistent state-level grouping**

```python
df['area'].unique()
```

**Verifying categorical columns (leak_status, body_type, confidence) are already clean**

```python
print(df['leak_status'].unique())
print(df['body_type'].unique())
print(df['confidence'].unique())
```

```python
df['state_clean'] = df['area'].str.split('(').str[0].str.strip()
```

```python
df['state_clean'].unique()
```

```python
df[df['area'].str.contains('Jharkhand \(arrest\)', na=False, regex=True)]
```

Note: one row ("Bihar (Patna) / Jharkhand (arrest)") spans two states and is simplified to Bihar.

**Saving cleaned dataset**

```python
df.to_csv('../data/processed/incidents_clean.csv', index=False)
```
